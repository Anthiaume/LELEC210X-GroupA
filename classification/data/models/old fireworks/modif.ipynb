```python
import pickle

model1 = pickle.load(open("model_Fireworks_v24_HF.pkl", "rb"))
model2 = pickle.load(open("model_fireworks_v25_AF.pkl", "rb"))

n_freqs1 = model1["params"]["keeped_frequencies"][1] - model1["params"]["keeped_frequencies"][0] + 1
n_freqs2 = model2["params"]["keeped_frequencies"][1] - model2["params"]["keeped_frequencies"][0] + 1

model1["params"]["n_freq"] = n_freqs1
model2["params"]["n_freq"] = n_freqs2

print(f"Model 1: {model1['model_name']} - Number of frequencies: {n_freqs1}")
print(f"Model 2: {model2['model_name']} - Number of frequencies: {n_freqs2}")

print(f"Model 1 params: {model1['params']}")
print(f"Model 2 params: {model2['params']}")

pickle.dump(model1, open("model_Fireworks_v24_HF.pkl", "wb"))
pickle.dump(model2, open("model_fireworks_v25_AF.pkl", "wb"))
```
